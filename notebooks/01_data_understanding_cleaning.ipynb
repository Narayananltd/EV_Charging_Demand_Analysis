```python

from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

# Locate the project folder
PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

# Locate the raw dataset
DATA_PATH = PROJECT_ROOT / "data" / "raw" / "ChargingRecords.csv"

print("Project folder:", PROJECT_ROOT)
print("Dataset path:", DATA_PATH)
print("Dataset exists:", DATA_PATH.exists())
```

```python

df = pd.read_csv(DATA_PATH)

print("DATA LOADING COMPLETED")
print("Dataset shape:", df.shape)

display(df.head())
```

```python

# Display column names
print("COLUMN NAMES:")
print(df.columns.tolist())

print("\nFIRST 5 ROWS:")
display(df.head())

print("\nDATASET INFORMATION:")
df.info()

print("\nMISSING VALUES:")
display(df.isnull().sum().to_frame(name="Missing Count"))

print("\nDUPLICATE ROWS:")
print("Total duplicate rows:", df.duplicated().sum())

print("\nNUMERICAL SUMMARY:")
display(df.describe().T)
```

```python

# 1. Investigate duration values
print("DURATION QUALITY CHECK")
print("Negative durations:", (df["Duration"] < 0).sum())
print("Zero durations:", (df["Duration"] == 0).sum())
print("Positive durations:", (df["Duration"] > 0).sum())

# 2. Inspect records with invalid durations
print("\nRECORDS WITH ZERO OR NEGATIVE DURATION")
display(
    df.loc[df["Duration"] <= 0,
           ["UserID", "StartDatetime", "EndDatetime", "Duration", "Demand"]]
      .head(10)
)

# 3. Check duplicate records
print("\nDUPLICATE CHECK")
print("Exact duplicate rows:", df.duplicated().sum())

# 4. Convert datetime columns safely for validation
start_dt = pd.to_datetime(df["StartDatetime"], errors="coerce")
end_dt = pd.to_datetime(df["EndDatetime"], errors="coerce")

print("\nDATETIME VALIDATION")
print("Invalid start datetimes:", start_dt.isna().sum())
print("Invalid end datetimes:", end_dt.isna().sum())
print("End before start:", (end_dt < start_dt).sum())

# 5. Check Demand
print("\nDEMAND VALIDATION")
print("Negative Demand:", (df["Demand"] < 0).sum())
print("Zero Demand:", (df["Demand"] == 0).sum())
```

```python

# Create a temporary inspection copy
inspection_df = df.copy()

inspection_df["StartDatetime"] = pd.to_datetime(
    inspection_df["StartDatetime"], errors="coerce"
)
inspection_df["EndDatetime"] = pd.to_datetime(
    inspection_df["EndDatetime"], errors="coerce"
)

# Calculate duration from timestamps
inspection_df["CalculatedDuration"] = (
    inspection_df["EndDatetime"] - inspection_df["StartDatetime"]
).dt.total_seconds() / 60

# Find records where end datetime is earlier than start datetime
suspicious_records = inspection_df[
    inspection_df["EndDatetime"] < inspection_df["StartDatetime"]
]

print("Records where end is before start:", len(suspicious_records))

display(
    suspicious_records[
        ["UserID", "StartDay", "StartTime", "EndDay", "EndTime",
         "StartDatetime", "EndDatetime", "Duration", "CalculatedDuration"]
    ]
)

# Compare stored and timestamp-calculated durations
print("\nDURATION COMPARISON")
print("Negative stored durations:", (inspection_df["Duration"] < 0).sum())
print("Zero stored durations:", (inspection_df["Duration"] == 0).sum())
print(
    "Records with duration mismatch:",
    (
        inspection_df["Duration"] - inspection_df["CalculatedDuration"]
    ).abs().gt(1).sum()
)
```

```python

# Create a temporary copy for investigation
check_df = df.copy()

check_df["StartDatetime"] = pd.to_datetime(
    check_df["StartDatetime"], errors="coerce"
)
check_df["EndDatetime"] = pd.to_datetime(
    check_df["EndDatetime"], errors="coerce"
)

# Calculate elapsed minutes from the available timestamps
check_df["CalculatedDuration"] = (
    check_df["EndDatetime"] - check_df["StartDatetime"]
).dt.total_seconds() / 60

check_df["DurationDifference"] = (
    check_df["Duration"] - check_df["CalculatedDuration"]
)

# Summary of differences
print("DURATION DIFFERENCE SUMMARY")
display(check_df["DurationDifference"].describe())

# Show 10 records with the largest absolute differences
print("\nLARGEST DURATION DIFFERENCES")
display(
    check_df.loc[
        check_df["DurationDifference"].abs().nlargest(10).index,
        [
            "StartDay", "StartTime", "EndDay", "EndTime",
            "Duration", "CalculatedDuration", "DurationDifference"
        ]
    ]
)

# Compare the separate date/time fields with combined datetime columns
print("\nZERO-DURATION EXAMPLES")
display(
    check_df.loc[
        check_df["Duration"] == 0,
        [
            "StartDay", "StartTime", "EndDay", "EndTime",
            "StartDatetime", "EndDatetime", "Duration", "Demand"
        ]
    ].head(10)
)
```

```python

# Work on a separate copy; keep the original df unchanged
time_check = df.copy()

# Reconstruct full timestamps using the date and time fields
time_check["StartFull"] = pd.to_datetime(
    time_check["StartDay"].astype(str) + " " +
    time_check["StartTime"].astype(str),
    errors="coerce"
)

time_check["EndFull"] = pd.to_datetime(
    time_check["EndDay"].astype(str) + " " +
    time_check["EndTime"].astype(str),
    errors="coerce"
)

# Calculate duration using the full timestamps
time_check["DurationFromFullTime"] = (
    time_check["EndFull"] - time_check["StartFull"]
).dt.total_seconds() / 60

# Compare stored duration with reconstructed duration
time_check["DurationDifferenceFull"] = (
    time_check["Duration"] - time_check["DurationFromFullTime"]
)

print("Invalid reconstructed start times:",
      time_check["StartFull"].isna().sum())
print("Invalid reconstructed end times:",
      time_check["EndFull"].isna().sum())
print("End before start:",
      (time_check["EndFull"] < time_check["StartFull"]).sum())
print("Negative stored durations:",
      (time_check["Duration"] < 0).sum())
print("Zero stored durations:",
      (time_check["Duration"] == 0).sum())
print("Duration mismatches over 1 minute:",
      time_check["DurationDifferenceFull"].abs().gt(1).sum())

print("\nLARGEST REMAINING MISMATCHES")
display(
    time_check.loc[
        time_check["DurationDifferenceFull"].abs().nlargest(10).index,
        [
            "StartDay", "StartTime", "EndDay", "EndTime",
            "Duration", "DurationFromFullTime",
            "DurationDifferenceFull"
        ]
    ]
)
```

```python

# Summarize the relationship between stored and calculated durations
valid_time = time_check[
    time_check["DurationFromFullTime"].notna()
].copy()

valid_time["AbsoluteDifference"] = (
    valid_time["Duration"] - valid_time["DurationFromFullTime"]
).abs()

print("TOTAL RECORDS:", len(valid_time))
print(
    "Exact matches (within 1 minute):",
    (valid_time["AbsoluteDifference"] <= 1).sum()
)
print(
    "Differences greater than 1 minute:",
    (valid_time["AbsoluteDifference"] > 1).sum()
)

# Compare stored duration with timestamp-based duration
print("\nCORRELATION BETWEEN DURATIONS:")
display(
    valid_time[["Duration", "DurationFromFullTime"]].corr()
)

# Inspect records with the largest duration differences
print("\nLARGEST MISMATCHES — INCLUDING DEMAND:")
display(
    valid_time.loc[
        valid_time["AbsoluteDifference"].nlargest(10).index,
        [
            "UserID", "ChargerID", "Location", "ChargerType",
            "StartFull", "EndFull", "Duration",
            "DurationFromFullTime", "Demand"
        ]
    ]
)
```

```python

# Flag records with duration mismatches
time_check["DurationMismatch"] = (
    time_check["DurationDifferenceFull"].abs() > 1
)

time_check["InvalidDuration"] = time_check["Duration"] < 0
time_check["ZeroDuration"] = time_check["Duration"] == 0

print("DURATION QUALITY BY LOCATION")
display(
    time_check.groupby("Location").agg(
        TotalRecords=("Duration", "size"),
        Mismatches=("DurationMismatch", "sum"),
        NegativeDurations=("InvalidDuration", "sum"),
        ZeroDurations=("ZeroDuration", "sum")
    ).sort_values("Mismatches", ascending=False)
)

print("\nDURATION QUALITY BY CHARGER TYPE")
display(
    time_check.groupby("ChargerType").agg(
        TotalRecords=("Duration", "size"),
        Mismatches=("DurationMismatch", "sum"),
        NegativeDurations=("InvalidDuration", "sum"),
        ZeroDurations=("ZeroDuration", "sum")
    )
)

print("\nDURATION DISTRIBUTION")
display(time_check["Duration"].describe())
```

```python

# Inspect all records with negative stored durations
negative_records = time_check.loc[
    time_check["Duration"] < 0,
    [
        "UserID", "ChargerID", "ChargerCompany",
        "Location", "ChargerType",
        "StartFull", "EndFull",
        "Duration", "DurationFromFullTime", "Demand"
    ]
].sort_values("Location")

print("Negative duration records:", len(negative_records))
display(negative_records)

# Summarize negative durations by location and charger type
print("\nNEGATIVE DURATIONS BY LOCATION AND CHARGER TYPE")
display(
    time_check.loc[time_check["Duration"] < 0]
    .groupby(["Location", "ChargerType"])
    .agg(
        RecordCount=("Duration", "size"),
        AverageDuration=("Duration", "mean"),
        AverageDemand=("Demand", "mean")
    )
)
```

```python

# Preserve the original dataset and create a working copy
df_clean = df.copy()

initial_rows = len(df_clean)

# 1. Remove exact duplicate rows
duplicate_count = int(df_clean.duplicated().sum())
df_clean = df_clean.drop_duplicates().copy()

# 2. Parse date and time columns
df_clean["StartDay"] = pd.to_datetime(
    df_clean["StartDay"], errors="coerce"
)
df_clean["EndDay"] = pd.to_datetime(
    df_clean["EndDay"], errors="coerce"
)

df_clean["StartDatetime"] = pd.to_datetime(
    df_clean["StartDay"].dt.strftime("%Y-%m-%d") + " " +
    df_clean["StartTime"].astype(str),
    errors="coerce"
)

df_clean["EndDatetime"] = pd.to_datetime(
    df_clean["EndDay"].dt.strftime("%Y-%m-%d") + " " +
    df_clean["EndTime"].astype(str),
    errors="coerce"
)

# 3. Remove records with invalid timestamps
invalid_datetime_mask = (
    df_clean["StartDatetime"].isna()
    | df_clean["EndDatetime"].isna()
)
invalid_datetime_count = int(invalid_datetime_mask.sum())
df_clean = df_clean.loc[~invalid_datetime_mask].copy()

# 4. Remove negative stored durations from the main cleaned dataset
negative_duration_count = int((df_clean["Duration"] < 0).sum())
df_clean = df_clean.loc[df_clean["Duration"] >= 0].copy()

# 5. Create useful time-based features
df_clean["StartHour"] = df_clean["StartDatetime"].dt.hour
df_clean["StartMonth"] = df_clean["StartDatetime"].dt.month
df_clean["StartWeekday"] = df_clean["StartDatetime"].dt.day_name()
df_clean["StartDate"] = df_clean["StartDatetime"].dt.date

# 6. Report cleaning results
print("INITIAL ROWS:", initial_rows)
print("DUPLICATES REMOVED:", duplicate_count)
print("INVALID DATETIME ROWS REMOVED:", invalid_datetime_count)
print("NEGATIVE DURATION ROWS REMOVED:", negative_duration_count)
print("FINAL ROWS:", len(df_clean))
print("ZERO-DURATION ROWS RETAINED:", (df_clean["Duration"] == 0).sum())
print("MISSING VALUES AFTER CLEANING:", int(df_clean.isna().sum().sum()))

display(df_clean.head())
```

```python

# Create the processed data folder
PROCESSED_PATH = PROJECT_ROOT / "data" / "processed"
PROCESSED_PATH.mkdir(parents=True, exist_ok=True)

# Save the cleaned dataset
cleaned_file = PROCESSED_PATH / "ChargingRecords_Cleaned.csv"
df_clean.to_csv(cleaned_file, index=False)

print("CLEANING PIPELINE COMPLETED SUCCESSFULLY")
print("Saved file:", cleaned_file)
print("Final dataset shape:", df_clean.shape)
```

```python

# Calculate session duration from timestamps (in minutes)
df_clean["DurationFromTimestamps"] = (
    df_clean["EndDatetime"] - df_clean["StartDatetime"]
).dt.total_seconds() / 60

# Compare stored and calculated duration
df_clean["DurationDifference"] = (
    df_clean["Duration"] - df_clean["DurationFromTimestamps"]
)

mismatch_mask = df_clean["DurationDifference"].abs() > 1

print("Total cleaned records:", len(df_clean))
print("Duration mismatches (> 1 minute):", int(mismatch_mask.sum()))
print("Records matching within 1 minute:", int((~mismatch_mask).sum()))

display(
    df_clean.loc[
        mismatch_mask,
        [
            "Location",
            "StartDatetime",
            "EndDatetime",
            "Duration",
            "DurationFromTimestamps",
            "DurationDifference",
        ],
    ].head(10)
)
```

```python

# Summarize duration mismatches by location
mismatch_summary = (
    df_clean.assign(DurationMismatch=mismatch_mask)
    .groupby("Location")
    .agg(
        TotalRecords=("DurationMismatch", "size"),
        MismatchRecords=("DurationMismatch", "sum"),
    )
    .reset_index()
)

mismatch_summary["MismatchPercentage"] = (
    mismatch_summary["MismatchRecords"]
    / mismatch_summary["TotalRecords"] * 100
).round(2)

mismatch_summary = mismatch_summary.sort_values(
    "MismatchRecords", ascending=False
)

display(mismatch_summary)

# Inspect the 10 largest duration discrepancies
largest_mismatches = df_clean.loc[
    mismatch_mask,
    [
        "Location",
        "ChargerType",
        "StartDatetime",
        "EndDatetime",
        "Duration",
        "DurationFromTimestamps",
        "DurationDifference",
    ],
].copy()

largest_mismatches["AbsoluteDifference"] = (
    largest_mismatches["DurationDifference"].abs()
)

display(
    largest_mismatches.sort_values(
        "AbsoluteDifference", ascending=False
    ).head(10)
)
```

```python

# Classify mismatches by their absolute size, in minutes
abs_diff = df_clean["DurationDifference"].abs()

severity_summary = pd.Series({
    "Mismatch > 1 minute": (abs_diff > 1).sum(),
    "Mismatch > 30 minutes": (abs_diff > 30).sum(),
    "Mismatch > 60 minutes": (abs_diff > 60).sum(),
    "Mismatch > 6 hours": (abs_diff > 360).sum(),
    "Mismatch > 24 hours": (abs_diff > 1440).sum(),
})

display(severity_summary.to_frame(name="RecordCount"))

# Check whether the stored duration or timestamp duration is larger
print(
    "Stored duration is shorter:",
    int((df_clean.loc[mismatch_mask, "DurationDifference"] < -1).sum())
)

print(
    "Stored duration is longer:",
    int((df_clean.loc[mismatch_mask, "DurationDifference"] > 1).sum())
)
```

```python

# Flag records with duration inconsistencies
df_clean["DurationMismatch"] = (
    df_clean["DurationDifference"].abs() > 1
).astype(int)

# Categorize mismatch severity
df_clean["DurationQuality"] = np.select(
    [
        df_clean["DurationDifference"].abs() <= 1,
        df_clean["DurationDifference"].abs() <= 30,
        df_clean["DurationDifference"].abs() <= 60,
        df_clean["DurationDifference"].abs() <= 360,
    ],
    [
        "Match",
        "Minor mismatch",
        "Moderate mismatch",
        "Major mismatch",
    ],
    default="Severe mismatch",
)

# Summarize the quality flags
print("Duration quality summary:")
display(
    df_clean["DurationQuality"]
    .value_counts()
    .rename_axis("DurationQuality")
    .to_frame("RecordCount")
)

print("Total records:", len(df_clean))
print("Missing values:", int(df_clean.isna().sum().sum()))
```
